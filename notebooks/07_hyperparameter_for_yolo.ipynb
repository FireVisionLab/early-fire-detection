Start from the freeze→unfreeze best weights (best F2 so far). Short genetic tune with model.tune (the genetic algorithm approach to hyperparameter optimization that is developed by ultralytics) then full retrain on the best hyp.

```python
import sys
from pathlib import Path
from ultralytics import YOLO
PROJECT_ROOT = Path("..").resolve()
sys.path.insert(0, str(PROJECT_ROOT))
from helpers.utils import set_seed, yolo_log_kwargs, f1_score, f2_score
set_seed(42)
DATA_YAML = PROJECT_ROOT / "data" / "dataset.yaml"
RUNS = PROJECT_ROOT / "models" / "runs"
START = RUNS / "yolo26n_unfreeze50" / "weights" / "best.pt"
```

```python
model = YOLO(str(START))
model
```

```python
tune_results = model.tune(
    data=str(DATA_YAML),
    epochs=25,
    iterations=30,
    optimizer="AdamW",
    plots=False,
    save=False,
    val=True,
    device=0,
    project=str(RUNS),
    name="yolo26n_tune",
    exist_ok=True,
)
```
