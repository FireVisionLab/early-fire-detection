```python
import sys
from pathlib import Path
ROOT = Path("/content/drive/MyDrive/early-fire-detection")
sys.path.insert(0, str(ROOT))
import yaml
from ultralytics import YOLO
```

```python
ROOT = Path("/content/drive/MyDrive/early-fire-detection")
DATA = ROOT / "data"
DATA_YAML = DATA / "dataset.yaml"
START = ROOT / "models" / "runs" / "yolo26n_unfreeze50" / "weights" / "best.pt"
RUNS = ROOT / "models" / "runs"
```

```python
cfg = yaml.safe_load(DATA_YAML.read_text())
cfg["path"] = str(DATA)
DATA_YAML.write_text(yaml.safe_dump(cfg, sort_keys=False))
```

We saw the model converges to the optima approximately at epoch 25. So we decided to tune models with epoch 25

```python
model = YOLO(str(START))
tune_results = model.tune(
    data=str(DATA_YAML),
    epochs=25,
    iterations=30,
    optimizer="AdamW",
    plots=False,
    save=False,
    val=True,
    device=0,
    project=str(RUNS),
    name="yolo26n_tune",
    exist_ok=True,
)
```
